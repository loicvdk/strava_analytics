```python
"""
Strava Zone Analysis - Pace Change Based Detection

This notebook loads Strava activity data and visualizes running zones
based on pace acceleration/deceleration patterns rather than absolute pace values.
"""

import polars as pl
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from datetime import timedelta
from pathlib import Path
from scipy.signal import savgol_filter

# File paths
ACTIVITIES_PATH = "/Users/loicvanderkelen/Documents/projects/strava/strava_analytics/kedro-strava-analytics/data/01_raw/strava_activities.csv"
STREAMS_PATH = "/Users/loicvanderkelen/Documents/projects/strava/strava_analytics/kedro-strava-analytics/data/02_intermediate/activity_streams.parquet"

# Pace change thresholds (athlete-independent)
PACE_CHANGE_THRESHOLDS = {
    # Pace acceleration thresholds (seconds/km change per data point)
    'threshold_acceleration': -60.0,  # Getting faster by >3 sec/km = likely threshold effort
    'recovery_deceleration': 60.0,    # Getting slower by >3 sec/km = likely recovery
    'steady_range': 20.0,             # Within ±1 sec/km = steady effort
    
    # Heart rate acceleration thresholds (bpm change per minute)
    'hr_acceleration_threshold': 10.0,  # HR rising by >3 bpm/min = effort increasing
    'hr_deceleration_threshold': -10.0, # HR dropping by >2 bpm/min = effort decreasing
    
    # Smoothing parameters
    'pace_smooth_window': 5,  # Window for pace smoothing (must be odd)
    'hr_smooth_window': 7,    # Window for HR smoothing (must be odd)
    
    # Minimum durations for valid efforts (seconds)
    'min_threshold_duration': 60,   # At least 1 minute for threshold
    'min_zone2_duration': 120,      # At least 2 minutes for Zone 2
}

# Zone definitions (only for HR - pace is now relative)
HR_ZONES = {
    'zone_1_max': 90,
    'zone_2_max': 130,
    'zone_3_max': 148,
    'zone_4_max': 167,
    'zone_5_max': 200,
}


def load_activities_and_streams():
    """
    Load activities and streams data.
    
    Returns
    -------
    tuple
        (activities_df, streams_df)
    """
    print("Loading data...")
    
    # Load activities
    activities = pl.read_csv(ACTIVITIES_PATH)
    print(f"✓ Loaded {len(activities)} activities")
    print(f"  Columns: {activities.columns[:5]}...")  # Show first 5 columns
    
    # Load streams
    streams = pl.read_parquet(STREAMS_PATH)
    print(f"✓ Loaded {len(streams)} stream records")
    print(f"  Columns: {streams.columns[:5]}...")
    
    return activities, streams


def get_latest_running_activity(activities, streams):
    """
    Get the latest running activity with streams data.
    
    Parameters
    ----------
    activities : pl.DataFrame
        Activities dataframe
    streams : pl.DataFrame
        Streams dataframe
        
    Returns
    -------
    tuple
        (activity_row, activity_streams)
    """
    # Filter for running activities
    running_activities = activities.filter(pl.col('type') == 'Run')
    
    if len(running_activities) == 0:
        raise ValueError("No running activities found")
    
    # Find date column
    date_col = None
    for col in ['start_date', 'start_date_local', 'date', 'start_time']:
        if col in running_activities.columns:
            date_col = col
            break
    
    if date_col is None:
        print("Available columns:", running_activities.columns)
        raise ValueError("Could not find date column in activities")
    
    # Sort by date and get the latest
    running_activities = running_activities.sort(date_col, descending=True)
    latest_activity = running_activities[2]
    
    # Get activity ID
    activity_id = latest_activity['id'][0]
    
    print(f"\n✓ Latest running activity:")
    print(f"  ID: {activity_id}")
    
    # Safely get activity name
    if 'name' in latest_activity.columns:
        print(f"  Name: {latest_activity['name'][0]}")
    
    print(f"  Date: {latest_activity[date_col][0]}")
    
    if 'distance' in latest_activity.columns:
        print(f"  Distance: {latest_activity['distance'][0] / 1000:.2f} km")
    
    # Get streams for this activity
    activity_streams = streams.filter(pl.col('activity_id') == activity_id)
    
    if len(activity_streams) == 0:
        raise ValueError(f"No streams found for activity {activity_id}")
    
    print(f"  Stream points: {len(activity_streams)}")
    
    return latest_activity, activity_streams


def smooth_data(data_array, window_size=11, poly_order=3):
    """
    Smooth data using Savitzky-Golay filter.
    
    Parameters
    ----------
    data_array : np.array
        Data to smooth
    window_size : int
        Size of smoothing window (must be odd)
    poly_order : int
        Polynomial order for fitting
        
    Returns
    -------
    np.array
        Smoothed data
    """
    if len(data_array) < window_size:
        window_size = len(data_array) if len(data_array) % 2 == 1 else len(data_array) - 1
        if window_size < 3:
            return data_array
    
    # Handle NaN values
    mask = ~np.isnan(data_array)
    if not mask.any():
        return data_array
    
    smoothed = data_array.copy()
    smoothed[mask] = savgol_filter(data_array[mask], window_size, min(poly_order, window_size - 1))
    
    return smoothed


def calculate_pace_changes_and_zones(streams_df, thresholds, hr_zones):
    """
    Calculate pace changes and classify zones based on acceleration patterns.
    
    Parameters
    ----------
    streams_df : pl.DataFrame
        Streams dataframe
    thresholds : dict
        Pace change thresholds
    hr_zones : dict
        Heart rate zone definitions
        
    Returns
    -------
    pl.DataFrame
        Enhanced streams with pace changes and zones
    """
    print("\n📊 Analyzing pace changes and zones...")
    
    # Convert to pandas for easier manipulation with scipy
    streams_df = streams_df.pivot(
        index=['activity_id', 'time_index'],
        columns='stream_type',
        values='value'
    )

    df = streams_df.to_pandas()
    
    # Find velocity column
    velocity_col = None
    for col in ['velocity_smooth', 'velocity', 'speed']:
        if col in df.columns:
            velocity_col = col
            break
    
    if velocity_col is None:
        raise ValueError("No velocity column found in streams")
    
    print(f"  Using velocity column: {velocity_col}")
    
    # Calculate pace
    df['pace_sec_per_km'] = np.where(
        df[velocity_col] > 0,
        1000 / df[velocity_col],
        np.nan
    )
    
    # Clip extreme outliers (likely GPS errors)
    df['pace_sec_per_km'] = df['pace_sec_per_km'].clip(lower=180, upper=600)
    
    # Smooth pace data
    print("  Smoothing pace data...")
    df['pace_smooth'] = smooth_data(
        df['pace_sec_per_km'].fillna(method='ffill').fillna(method='bfill').values,
        window_size=thresholds['pace_smooth_window'],
        poly_order=3
    )
    
    # Calculate pace acceleration (negative = getting faster)
    df['pace_change'] = df['pace_smooth'].diff()
    df['pace_change_smooth'] = smooth_data(
        df['pace_change'].fillna(0).values,
        window_size=5,
        poly_order=2
    )
    
    # Find HR column
    hr_col = None
    for col in ['heartrate', 'heart_rate', 'hr']:
        if col in df.columns:
            hr_col = col
            break
    
    has_hr = hr_col is not None
    
    if has_hr:
        print(f"  Using heart rate column: {hr_col}")
        df['heartrate'] = df[hr_col]
        
        # Smooth HR data
        print("  Smoothing heart rate data...")
        df['hr_smooth'] = smooth_data(
            df['heartrate'].fillna(method='ffill').fillna(method='bfill').values,
            window_size=thresholds['hr_smooth_window'],
            poly_order=3
        )
        
        # Calculate HR acceleration (bpm per minute)
        df['hr_change'] = df['hr_smooth'].diff() * 60  # Convert to per minute
        df['hr_change_smooth'] = smooth_data(
            df['hr_change'].fillna(0).values,
            window_size=7,
            poly_order=2
        )
        
        # Classify HR zones
        df['hr_zone'] = pd.cut(
            df['hr_smooth'],
            bins=[0, hr_zones['zone_1_max'], hr_zones['zone_2_max'], 
                  hr_zones['zone_3_max'], hr_zones['zone_4_max'], hr_zones['zone_5_max'], 300],
            labels=['Zone 1', 'Zone 2', 'Zone 3', 'Zone 4', 'Zone 5', 'Max']
        )
    
    # Classify effort based on pace changes
    print("  Classifying efforts based on pace changes...")
    
    df['effort_pattern'] = 'Steady'
    df['is_accelerating'] = df['pace_change_smooth'] < thresholds['threshold_acceleration']
    df['is_decelerating'] = df['pace_change_smooth'] > thresholds['recovery_deceleration']
    df['is_steady'] = np.abs(df['pace_change_smooth']) <= thresholds['steady_range']
    
    # Identify threshold efforts (sustained fast pace with HR response)
    df['is_threshold'] = False
    df['is_zone2'] = False
    
    # Find periods where pace is stable (not changing much) but fast
    # We'll use a rolling window to detect sustained efforts
    window_size = 30  # 30 seconds
    
    # Calculate rolling statistics
    df['pace_rolling_std'] = df['pace_smooth'].rolling(window=window_size, center=True).std()
    df['pace_rolling_mean'] = df['pace_smooth'].rolling(window=window_size, center=True).mean()
    
    if has_hr:
        df['hr_rolling_mean'] = df['hr_smooth'].rolling(window=window_size, center=True).mean()
        
        # Threshold effort: HR in zones 3-4 with relatively steady pace
        df['is_threshold'] = (
            (df['hr_rolling_mean'] >= hr_zones['zone_3_max']) &
            (df['hr_rolling_mean'] <= hr_zones['zone_4_max']) &
            (df['pace_rolling_std'] < 10)  # Pace variation less than 10 sec/km
        )
        
        # Zone 2 effort: HR in zone 2 with steady pace
        df['is_zone2'] = (
            (df['hr_rolling_mean'] > hr_zones['zone_1_max']) &
            (df['hr_rolling_mean'] <= hr_zones['zone_2_max']) &
            (df['pace_rolling_std'] < 15)  # Allow more variation in Zone 2
        )
        
        # Apply lag compensation: extend threshold backwards when pace increases
        df = apply_lag_compensation(df, thresholds)
    else:
        print("  ⚠️  No HR data available - using pace patterns only")
        # Without HR, we can only use pace stability
        # Threshold = fast sustained effort
        # Zone 2 = moderate sustained effort
        percentile_50 = df['pace_smooth'].quantile(0.5)
        percentile_30 = df['pace_smooth'].quantile(0.3)
        
        df['is_threshold'] = (
            (df['pace_smooth'] <= percentile_30) &
            (df['pace_rolling_std'] < 10)
        )
        
        df['is_zone2'] = (
            (df['pace_smooth'] <= percentile_50) &
            (df['pace_smooth'] > percentile_30) &
            (df['pace_rolling_std'] < 15)
        )
    
    # Create combined zone classification
    df['combined_zone'] = 'Easy/Recovery'
    df.loc[df['is_zone2'], 'combined_zone'] = 'Zone 2'
    df.loc[df['is_threshold'], 'combined_zone'] = 'Threshold'
    df.loc[df['is_accelerating'], 'effort_pattern'] = 'Accelerating'
    df.loc[df['is_decelerating'], 'effort_pattern'] = 'Decelerating'
    
    print(f"✓ Zone classification complete")
    print(f"  Threshold points: {df['is_threshold'].sum()}")
    print(f"  Zone 2 points: {df['is_zone2'].sum()}")
    
    return pl.from_pandas(df)


def apply_lag_compensation(df, thresholds):
    """
    Apply HR lag compensation by extending zones backwards.
    
    Parameters
    ----------
    df : pd.DataFrame
        DataFrame with initial zone classification
    thresholds : dict
        Threshold configuration
        
    Returns
    -------
    pd.DataFrame
        DataFrame with lag-compensated zones
    """
    print("  Applying HR lag compensation...")
    
    # Find transitions to threshold
    threshold_starts = df.index[df['is_threshold'] & ~df['is_threshold'].shift(1, fill_value=False)].tolist()
    
    lag_seconds = 45  # Typical HR lag
    
    for start_idx in threshold_starts:
        # Look backwards up to lag_seconds
        lookback_start = max(0, start_idx - lag_seconds)
        
        # Check if pace was already fast before HR caught up
        lookback_section = df.iloc[lookback_start:start_idx]
        
        # If pace was increasing (getting faster) before the threshold, include it
        if len(lookback_section) > 0:
            pace_was_increasing = lookback_section['pace_change_smooth'] < -1.0
            
            # Extend threshold classification backwards where pace was increasing
            for idx in lookback_section.index[pace_was_increasing]:
                df.at[idx, 'is_threshold'] = True
                df.at[idx, 'combined_zone'] = 'Threshold'
    
    # Similar logic for threshold ends
    threshold_ends = df.index[~df['is_threshold'] & df['is_threshold'].shift(1, fill_value=False)].tolist()
    
    for end_idx in threshold_ends:
        # Look forward to see if HR stays elevated while pace drops
        lookforward_end = min(len(df), end_idx + lag_seconds // 2)
        
        lookforward_section = df.iloc[end_idx:lookforward_end]
        
        if len(lookforward_section) > 0:
            # If HR is still elevated but pace dropped, this is recovery not threshold
            hr_still_high = lookforward_section['hr_smooth'] > HR_ZONES['zone_3_max']
            pace_dropped = lookforward_section['pace_change_smooth'] > 2.0
            
            # Remove threshold classification where it's clearly recovery
            for idx in lookforward_section.index[hr_still_high & pace_dropped]:
                df.at[idx, 'is_threshold'] = False
                df.at[idx, 'combined_zone'] = 'Recovery'
    
    return df


def format_pace(seconds_per_km):
    """Format pace to MM:SS/km string."""
    if np.isnan(seconds_per_km) or seconds_per_km == 0:
        return "0:00"
    minutes = int(seconds_per_km // 60)
    seconds = int(seconds_per_km % 60)
    return f"{minutes}:{seconds:02d}"


def plot_pace_changes_analysis(streams_df, activity_name="Running Activity"):
    """
    Plot pace and pace changes with zone coloring.
    
    Parameters
    ----------
    streams_df : pl.DataFrame
        Streams with pace change data
    activity_name : str
        Activity name
    """
    df = streams_df.to_pandas()
    
    # Create figure with 3 subplots
    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(16, 12), sharex=True)
    
    # Find time column
    time_col = None
    for col in ['time', 'elapsed_time', 'timestamp']:
        if col in df.columns:
            time_col = col
            break
    
    if time_col is None:
        raise ValueError("No time column found")
    
    # Convert time to minutes
    time_minutes = df[time_col] / 60
    
    # Zone colors
    zone_colors = {
        'Zone 2': '#4CAF50',
        'Threshold': '#FF9800',
        'Easy/Recovery': '#2196F3',
        'Recovery': '#81C784',
        'Other': '#9E9E9E'
    }
    
    # Plot 1: Pace with zone backgrounds
    if 'pace_smooth' in df.columns and 'combined_zone' in df.columns:
        # Add zone backgrounds
        current_zone = None
        zone_start = 0
        
        for i in range(len(df)):
            zone = df.iloc[i]['combined_zone']
            
            if zone != current_zone:
                if current_zone is not None and current_zone in zone_colors:
                    ax1.axvspan(zone_start, time_minutes.iloc[i], 
                               alpha=0.2, color=zone_colors[current_zone])
                current_zone = zone
                zone_start = time_minutes.iloc[i]
        
        if current_zone in zone_colors:
            ax1.axvspan(zone_start, time_minutes.iloc[-1], 
                       alpha=0.2, color=zone_colors[current_zone])
        
        # Plot pace
        ax1.plot(time_minutes, df['pace_smooth'], 'k-', linewidth=2, label='Pace (smoothed)')
        ax1.invert_yaxis()
        
        pace_ticks = ax1.get_yticks()
        ax1.set_yticklabels([format_pace(p) for p in pace_ticks])
        
        ax1.set_ylabel('Pace (min:sec/km)', fontsize=12, fontweight='bold')
        ax1.set_title(f'Pace Analysis with Zone Detection: {activity_name}', 
                     fontsize=14, fontweight='bold')
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='upper right')
    
    # Plot 2: Pace Change Rate
    if 'pace_change_smooth' in df.columns:
        ax2.plot(time_minutes, df['pace_change_smooth'], 'b-', linewidth=2, label='Pace Change Rate')
        ax2.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
        ax2.axhline(y=PACE_CHANGE_THRESHOLDS['threshold_acceleration'], 
                   color='orange', linestyle='--', alpha=0.7, label='Acceleration Threshold')
        ax2.axhline(y=PACE_CHANGE_THRESHOLDS['recovery_deceleration'], 
                   color='green', linestyle='--', alpha=0.7, label='Deceleration Threshold')
        
        # Add zone backgrounds
        current_zone = None
        zone_start = 0
        
        for i in range(len(df)):
            zone = df.iloc[i]['combined_zone']
            
            if zone != current_zone:
                if current_zone is not None and current_zone in zone_colors:
                    ax2.axvspan(zone_start, time_minutes.iloc[i], 
                               alpha=0.15, color=zone_colors[current_zone])
                current_zone = zone
                zone_start = time_minutes.iloc[i]
        
        if current_zone in zone_colors:
            ax2.axvspan(zone_start, time_minutes.iloc[-1], 
                       alpha=0.15, color=zone_colors[current_zone])
        
        ax2.set_ylabel('Pace Change (sec/km)', fontsize=12, fontweight='bold')
        ax2.set_title('Pace Acceleration/Deceleration', fontsize=14, fontweight='bold')
        ax2.grid(True, alpha=0.3)
        ax2.legend(loc='upper right')
    
    # Plot 3: Heart Rate (if available)
    if 'hr_smooth' in df.columns:
        ax3.plot(time_minutes, df['hr_smooth'], 'r-', linewidth=2, label='Heart Rate')
        
        # Add HR zone lines
        for zone_name, hr_value in [('Zone 2 Max', HR_ZONES['zone_2_max']),
                                     ('Zone 3 Max', HR_ZONES['zone_3_max']),
                                     ('Zone 4 Max', HR_ZONES['zone_4_max'])]:
            ax3.axhline(y=hr_value, color='gray', linestyle='--', alpha=0.5, linewidth=1)
            ax3.text(time_minutes.iloc[-1], hr_value, f' {zone_name}', 
                    va='center', fontsize=9, alpha=0.7)
        
        # Add zone backgrounds
        current_zone = None
        zone_start = 0
        
        for i in range(len(df)):
            zone = df.iloc[i]['combined_zone']
            
            if zone != current_zone:
                if current_zone is not None and current_zone in zone_colors:
                    ax3.axvspan(zone_start, time_minutes.iloc[i], 
                               alpha=0.15, color=zone_colors[current_zone])
                current_zone = zone
                zone_start = time_minutes.iloc[i]
        
        if current_zone in zone_colors:
            ax3.axvspan(zone_start, time_minutes.iloc[-1], 
                       alpha=0.15, color=zone_colors[current_zone])
        
        ax3.set_ylabel('Heart Rate (bpm)', fontsize=12, fontweight='bold')
        ax3.set_xlabel('Time (minutes)', fontsize=12, fontweight='bold')
        ax3.set_title('Heart Rate with Lag-Compensated Zones', fontsize=14, fontweight='bold')
        ax3.grid(True, alpha=0.3)
        ax3.legend(loc='upper right')
    
    # Create zone legend
    zone_patches = [mpatches.Patch(color=color, label=zone, alpha=0.5) 
                   for zone, color in zone_colors.items()]
    
    fig.legend(handles=zone_patches, loc='upper right', bbox_to_anchor=(0.99, 0.99),
              title='Training Zones', framealpha=0.9, fontsize=10)
    
    plt.tight_layout()
    plt.savefig('pace_change_analysis.png', dpi=300, bbox_inches='tight')
    print("\n💾 Plot saved as 'pace_change_analysis.png'")
    plt.show()


def print_zone_summary(streams_df):
    """Print summary of time in zones."""
    df = streams_df.to_pandas()
    
    if 'combined_zone' not in df.columns:
        print("No zone data available")
        return
    
    time_col = None
    for col in ['time', 'elapsed_time']:
        if col in df.columns:
            time_col = col
            break
    
    if time_col is None:
        return
    
    total_time = df[time_col].max()
    
    print("\n" + "="*70)
    print("ZONE DISTRIBUTION SUMMARY (Based on Pace Changes)")
    print("="*70)
    
    zone_stats = df.groupby('combined_zone').agg({
        time_col: 'count'
    }).rename(columns={time_col: 'count'})
    
    for zone, row in zone_stats.iterrows():
        count = row['count']
        percentage = (count / len(df)) * 100
        time_in_zone = (count / len(df)) * total_time
        
        print(f"\n{zone}:")
        print(f"  Time: {timedelta(seconds=int(time_in_zone))}")
        print(f"  Percentage: {percentage:.1f}%")
        print(f"  Data points: {count}")


def main():
    """Main execution function."""
    try:
        # Load data
        activities, streams = load_activities_and_streams()
        
        # Get latest running activity
        activity, activity_streams = get_latest_running_activity(activities, streams)
        
        # Calculate pace changes and zones
        activity_streams = calculate_pace_changes_and_zones(
            activity_streams, 
            PACE_CHANGE_THRESHOLDS,
            HR_ZONES
        )
        
        # Get activity name
        activity_name = "Running Activity"
        if 'name' in activity.columns:
            activity_name = activity['name'][0]
        
        # Plot
        plot_pace_changes_analysis(activity_streams, activity_name)
        
        # Print summary
        print_zone_summary(activity_streams)
        
        print("\n✅ Analysis complete!")
        
        return activity_streams
        
    except Exception as e:
        print(f"\n❌ Error: {e}")
        import traceback
        traceback.print_exc()
        return None


result = main()
result
```

```python
"""
Strava Zone Analysis - Simple 3-Zone Classification

This notebook classifies running into 3 zones:
1. Recovery (Zone 1-2)
2. Threshold (Zone 3-4) 
3. Above Threshold (Zone 5+)

Uses heart rate with lag compensation based on pace changes.
"""

import polars as pl
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from datetime import timedelta
from scipy.signal import savgol_filter

# File paths
ACTIVITIES_PATH = "/Users/loicvanderkelen/Documents/projects/strava/strava_analytics/kedro-strava-analytics/data/01_raw/strava_activities.csv"
STREAMS_PATH = "/Users/loicvanderkelen/Documents/projects/strava/strava_analytics/kedro-strava-analytics/data/02_intermediate/activity_streams.parquet"

# Simple 3-zone heart rate definitions (customize for athlete)
HR_ZONES = {
    'recovery_max': 130,      # Below this = Recovery (Zone 1-2)
    'threshold_max': 176,     # 155-183 = Threshold (Zone 3-4)
    # Above 183 = Above Threshold (Zone 5+)
}

# Lag compensation settings
LAG_SETTINGS = {
    'hr_lag_seconds': 45,           # Typical HR lag time
    'min_pace_increase': -2.0,      # sec/km - getting faster (negative = faster)
    'min_pace_decrease': 60.0,       # sec/km - getting slower (positive = slower)
    'smoothing_window': 11,         # Window for smoothing (must be odd)
}


def load_activities_and_streams():
    """Load activities and streams data."""
    print("Loading data...")
    
    activities = pl.read_csv(ACTIVITIES_PATH)
    print(f"✓ Loaded {len(activities)} activities")
    
    streams = pl.read_parquet(STREAMS_PATH)
    print(f"✓ Loaded {len(streams)} stream records")
    
    return activities, streams


def get_latest_running_activity(activities, streams):
    """Get the latest running activity with streams data."""
    # Filter for running activities
    running_activities = activities.filter(pl.col('type') == 'Run')
    
    if len(running_activities) == 0:
        raise ValueError("No running activities found")
    
    # Find date column
    date_col = None
    for col in ['start_date', 'start_date_local', 'date', 'start_time']:
        if col in running_activities.columns:
            date_col = col
            break
    
    if date_col is None:
        raise ValueError("Could not find date column")
    
    # Sort and get latest
    running_activities = running_activities.sort(date_col, descending=True)
    latest_activity = running_activities[2]
    
    activity_id = latest_activity['id'][0]
    
    print(f"\n✓ Latest running activity:")
    print(f"  ID: {activity_id}")
    
    if 'name' in latest_activity.columns:
        print(f"  Name: {latest_activity['name'][0]}")
    
    print(f"  Date: {latest_activity[date_col][0]}")
    
    if 'distance' in latest_activity.columns:
        print(f"  Distance: {latest_activity['distance'][0] / 1000:.2f} km")
    
    # Get streams and pivot
    activity_streams = streams.filter(pl.col('activity_id') == activity_id)
    
    if len(activity_streams) == 0:
        raise ValueError(f"No streams found for activity {activity_id}")
    
    print(f"\n📊 Pivoting stream data...")
    activity_streams = activity_streams.pivot(
        index=['activity_id', 'time_index'],
        columns='stream_type',
        values='value'
    ).sort('time_index')

    
    print(f"  Stream points: {len(activity_streams)}")
    print(f"  Available streams: {[col for col in activity_streams.columns if col not in ['activity_id', 'time']]}")
    
    return latest_activity, activity_streams


def smooth_data(data_array, window_size=11):
    """Smooth data using Savitzky-Golay filter."""
    if len(data_array) < window_size:
        window_size = len(data_array) if len(data_array) % 2 == 1 else len(data_array) - 1
        if window_size < 3:
            return data_array
    
    # Handle NaN values
    mask = ~np.isnan(data_array)
    if not mask.any():
        return data_array
    
    smoothed = data_array.copy()
    smoothed[mask] = savgol_filter(data_array[mask], window_size, 3)
    
    return smoothed


def classify_three_zones(streams_df, hr_zones, lag_settings):
    """
    Classify activity into 3 zones with HR lag compensation.
    
    Parameters
    ----------
    streams_df : pl.DataFrame
        Streams dataframe
    hr_zones : dict
        Heart rate zone thresholds
    lag_settings : dict
        Lag compensation settings
        
    Returns
    -------
    pl.DataFrame
        Enhanced streams with zone classifications
    """
    print("\n📊 Classifying into 3 zones...")
    
    # Convert to pandas for easier manipulation
    df = streams_df.to_pandas()
    
    # Find required columns
    velocity_col = None
    for col in ['velocity_smooth', 'velocity', 'speed']:
        if col in df.columns:
            velocity_col = col
            break
    
    hr_col = None
    for col in ['heartrate', 'heart_rate', 'hr']:
        if col in df.columns:
            hr_col = col
            break
    
    if velocity_col is None:
        raise ValueError("No velocity column found")
    
    if hr_col is None:
        raise ValueError("No heart rate column found - need HR data for zone classification")
    
    print(f"  Using velocity: {velocity_col}")
    print(f"  Using heart rate: {hr_col}")
    
    # Calculate pace
    df['pace_sec_per_km'] = np.where(
        df[velocity_col] > 0,
        1000 / df[velocity_col],
        np.nan
    )
    
    # Clip extreme outliers
    df['pace_sec_per_km'] = df['pace_sec_per_km'].clip(lower=180, upper=600)
    
    # Smooth pace and HR
    print("  Smoothing data...")
    df['pace_smooth'] = smooth_data(
        df['pace_sec_per_km'].fillna(method='ffill').fillna(method='bfill').values,
        window_size=lag_settings['smoothing_window']
    )
    
    df['hr_smooth'] = smooth_data(
        df[hr_col].fillna(method='ffill').fillna(method='bfill').values,
        window_size=lag_settings['smoothing_window']
    )
    
    # Calculate pace change (for lag compensation)
    df['pace_change'] = df['pace_smooth'].diff()
    
    # Step 1: Basic HR-based zone classification
    print("  Applying basic HR zone classification...")
    df['zone_basic'] = 'Recovery'
    df.loc[df['hr_smooth'] > hr_zones['recovery_max'], 'zone_basic'] = 'Threshold'
    df.loc[df['hr_smooth'] > hr_zones['threshold_max'], 'zone_basic'] = 'Above Threshold'
    
    # Step 2: Apply lag compensation
    print("  Applying HR lag compensation...")
    df['zone_compensated'] = df['zone_basic'].copy()
    
    lag_samples = lag_settings['hr_lag_seconds']
    
    # Find transitions INTO threshold or above threshold
    threshold_transitions = df.index[
        (df['zone_basic'].isin(['Threshold', 'Above Threshold'])) & 
        (~df['zone_basic'].shift(1, fill_value='Recovery').isin(['Threshold', 'Above Threshold']))
    ].tolist()
    
    for transition_idx in threshold_transitions:
        # Look backwards for pace increases
        lookback_start = max(0, transition_idx - lag_samples)
        lookback_section = df.iloc[lookback_start:transition_idx]
        
        if len(lookback_section) == 0:
            continue
        
        # Find where pace started increasing (getting faster = negative change)
        pace_increasing = lookback_section['pace_change'] < lag_settings['min_pace_increase']
        
        if pace_increasing.any():
            # Find the first point where pace started increasing
            first_increase_idx = lookback_section.index[pace_increasing][0]
            
            # Get the zone we're transitioning into
            target_zone = df.loc[transition_idx, 'zone_basic']
            
            # Extend the zone backwards to where pace started increasing
            df.loc[first_increase_idx:transition_idx, 'zone_compensated'] = target_zone
    
    # Find transitions OUT OF threshold/above threshold
    recovery_transitions = df.index[
        (df['zone_basic'] == 'Recovery') & 
        (df['zone_basic'].shift(1, fill_value='Recovery').isin(['Threshold', 'Above Threshold']))
    ].tolist()
    
    for transition_idx in recovery_transitions:
        # Look backwards to see where pace actually started decreasing
        lookback_start = max(0, transition_idx - lag_samples // 2)
        lookback_section = df.iloc[lookback_start:transition_idx]
        
        if len(lookback_section) == 0:
            continue
        
        # Find where pace started decreasing (getting slower = positive change)
        pace_decreasing = lookback_section['pace_change'] > lag_settings['min_pace_decrease']
        
        if pace_decreasing.any():
            # Find the first point where pace started slowing
            first_decrease_idx = lookback_section.index[pace_decreasing][0]
            
            # Change everything from that point to Recovery
            df.loc[first_decrease_idx:transition_idx, 'zone_compensated'] = 'Recovery'
    
    # Calculate statistics
    zone_counts = df['zone_compensated'].value_counts()
    print("\n  Zone distribution:")
    for zone in ['Recovery', 'Threshold', 'Above Threshold']:
        if zone in zone_counts.index:
            count = zone_counts[zone]
            percentage = (count / len(df)) * 100
            print(f"    {zone}: {count} points ({percentage:.1f}%)")
    
    # Compare with basic classification
    print("\n  Lag compensation impact:")
    basic_threshold = (df['zone_basic'] == 'Threshold').sum()
    comp_threshold = (df['zone_compensated'] == 'Threshold').sum()
    print(f"    Threshold: {basic_threshold} → {comp_threshold} points ({comp_threshold - basic_threshold:+d})")
    
    basic_above = (df['zone_basic'] == 'Above Threshold').sum()
    comp_above = (df['zone_compensated'] == 'Above Threshold').sum()
    print(f"    Above Threshold: {basic_above} → {comp_above} points ({comp_above - basic_above:+d})")
    
    return pl.from_pandas(df)


def format_pace(seconds_per_km):
    """Format pace to MM:SS/km string."""
    if np.isnan(seconds_per_km) or seconds_per_km == 0:
        return "0:00"
    minutes = int(seconds_per_km // 60)
    seconds = int(seconds_per_km % 60)
    return f"{minutes}:{seconds:02d}"


def plot_three_zone_analysis(streams_df, activity_name="Running Activity"):
    """
    Plot pace and HR with 3-zone classification.
    
    Parameters
    ----------
    streams_df : pl.DataFrame
        Streams with zone data
    activity_name : str
        Activity name
    """
    df = streams_df.to_pandas()
    
    # Create figure with 2 subplots
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 10), sharex=True)
    
    # Time in minutes
    time_minutes = df['time'] / 60
    
    # Zone colors - simple and clear
    zone_colors = {
        'Recovery': '#4CAF50',           # Green
        'Threshold': '#FF9800',          # Orange
        'Above Threshold': '#F44336',    # Red
    }
    
    # Helper function to add zone backgrounds
    def add_zone_backgrounds(ax, df, time_minutes, zone_col):
        current_zone = None
        zone_start = 0
        
        for i in range(len(df)):
            zone = df.iloc[i][zone_col]
            
            if zone != current_zone:
                if current_zone is not None and current_zone in zone_colors:
                    ax.axvspan(zone_start, time_minutes.iloc[i], 
                              alpha=0.25, color=zone_colors[current_zone])
                current_zone = zone
                zone_start = time_minutes.iloc[i]
        
        # Fill last zone
        if current_zone in zone_colors:
            ax.axvspan(zone_start, time_minutes.iloc[-1], 
                      alpha=0.25, color=zone_colors[current_zone])
    
    # Plot 1: Pace with compensated zones
    if 'pace_smooth' in df.columns and 'zone_compensated' in df.columns:
        add_zone_backgrounds(ax1, df, time_minutes, 'zone_compensated')
        
        # Plot pace
        ax1.plot(time_minutes, df['pace_smooth'], 'k-', linewidth=2.5, 
                label='Pace', zorder=10)
        ax1.invert_yaxis()
        
        # Format y-axis
        pace_ticks = ax1.get_yticks()
        ax1.set_yticklabels([format_pace(p) for p in pace_ticks])
        
        ax1.set_ylabel('Pace (min/km)', fontsize=13, fontweight='bold')
        ax1.set_title(f'3-Zone Analysis (Lag Compensated): {activity_name}', 
                     fontsize=15, fontweight='bold')
        ax1.grid(True, alpha=0.3, zorder=0)
        ax1.legend(loc='upper right', fontsize=11)
    
    # Plot 2: Heart Rate with compensated zones
    if 'hr_smooth' in df.columns and 'zone_compensated' in df.columns:
        add_zone_backgrounds(ax2, df, time_minutes, 'zone_compensated')
        
        # Plot HR
        ax2.plot(time_minutes, df['hr_smooth'], 'r-', linewidth=2.5, 
                label='Heart Rate', zorder=10)
        
        # Add zone threshold lines
        ax2.axhline(y=HR_ZONES['recovery_max'], color='gray', 
                   linestyle='--', alpha=0.6, linewidth=1.5, zorder=5)
        ax2.text(time_minutes.iloc[-1] * 0.02, HR_ZONES['recovery_max'], 
                f" Recovery Max ({HR_ZONES['recovery_max']} bpm)", 
                va='bottom', fontsize=10, alpha=0.8)
        
        ax2.axhline(y=HR_ZONES['threshold_max'], color='gray', 
                   linestyle='--', alpha=0.6, linewidth=1.5, zorder=5)
        ax2.text(time_minutes.iloc[-1] * 0.02, HR_ZONES['threshold_max'], 
                f" Threshold Max ({HR_ZONES['threshold_max']} bpm)", 
                va='bottom', fontsize=10, alpha=0.8)
        
        ax2.set_ylabel('Heart Rate (bpm)', fontsize=13, fontweight='bold')
        ax2.set_xlabel('Time (minutes)', fontsize=13, fontweight='bold')
        ax2.set_title('Heart Rate with 3-Zone Classification', 
                     fontsize=15, fontweight='bold')
        ax2.grid(True, alpha=0.3, zorder=0)
        ax2.legend(loc='upper right', fontsize=11)
    
    # Create zone legend
    zone_patches = [
        mpatches.Patch(color=zone_colors['Recovery'], label='Recovery (Z1-2)', alpha=0.6),
        mpatches.Patch(color=zone_colors['Threshold'], label='Threshold (Z3-4)', alpha=0.6),
        mpatches.Patch(color=zone_colors['Above Threshold'], label='Above Threshold (Z5+)', alpha=0.6)
    ]
    
    fig.legend(handles=zone_patches, loc='upper right', 
              bbox_to_anchor=(0.98, 0.98),
              title='Training Zones', framealpha=0.95, fontsize=11,
              title_fontsize=12)
    
    plt.tight_layout()
    plt.savefig('three_zone_analysis.png', dpi=300, bbox_inches='tight')
    print("\n💾 Plot saved as 'three_zone_analysis.png'")
    plt.show()


def print_zone_summary(streams_df):
    """Print summary of time in 3 zones."""
    df = streams_df.to_pandas()
    
    if 'zone_compensated' not in df.columns:
        print("No zone data available")
        return
    
    total_time = df['time'].max()
    
    print("\n" + "="*70)
    print("3-ZONE DISTRIBUTION SUMMARY")
    print("="*70)
    
    for zone in ['Recovery', 'Threshold', 'Above Threshold']:
        count = (df['zone_compensated'] == zone).sum()
        percentage = (count / len(df)) * 100
        time_in_zone = (count / len(df)) * total_time
        
        print(f"\n{zone}:")
        print(f"  Time: {timedelta(seconds=int(time_in_zone))}")
        print(f"  Percentage: {percentage:.1f}%")
        print(f"  Data points: {count}")
    
    # Calculate training stress metrics
    print("\n" + "="*70)
    print("TRAINING METRICS")
    print("="*70)
    
    recovery_pct = (df['zone_compensated'] == 'Recovery').sum() / len(df) * 100
    threshold_pct = (df['zone_compensated'] == 'Threshold').sum() / len(df) * 100
    above_pct = (df['zone_compensated'] == 'Above Threshold').sum() / len(df) * 100
    
    # Polarization index (80/20 rule: should have ~80% easy, ~20% hard)
    easy_hard_ratio = recovery_pct / (threshold_pct + above_pct) if (threshold_pct + above_pct) > 0 else 0
    
    print(f"\nPolarization Index: {easy_hard_ratio:.2f}")
    print(f"  (Optimal is ~4.0 for 80/20 training)")
    print(f"  Current split: {recovery_pct:.0f}% easy / {threshold_pct + above_pct:.0f}% hard")
    
    if easy_hard_ratio > 5:
        print("  → Very polarized training (lots of easy running)")
    elif easy_hard_ratio > 3:
        print("  → Well-polarized training (good 80/20 split)")
    elif easy_hard_ratio > 1.5:
        print("  → Moderate polarization")
    else:
        print("  → Low polarization (too much moderate/hard intensity)")


def main():
    """Main execution function."""
    try:
        # Load data
        activities, streams = load_activities_and_streams()
        
        # Get latest running activity
        activity, activity_streams = get_latest_running_activity(activities, streams)
        
        # Classify into 3 zones
        activity_streams = classify_three_zones(
            activity_streams,
            HR_ZONES,
            LAG_SETTINGS
        )
        
        # Get activity name
        activity_name = "Running Activity"
        if 'name' in activity.columns:
            activity_name = activity['name'][0]
        
        # Plot
        plot_three_zone_analysis(activity_streams, activity_name)
        
        # Print summary
        print_zone_summary(activity_streams)
        
        print("\n✅ Analysis complete!")
        
        return activity_streams
        
    except Exception as e:
        print(f"\n❌ Error: {e}")
        import traceback
        traceback.print_exc()
        return None


if __name__ == "__main__":
    result = main()
```
